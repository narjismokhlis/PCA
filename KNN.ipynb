This is my first implementation of the KNN algorithm. 

```python
def KNN(K, X, y, xtest):
    
    N, D = X.shape #training dataset
    N = y.shape[0] #its corresponding labels
    D = xtest.shape[0] #testing dataset
    
    #pair-wise distances 
    d = X - xtest 
    distance_matrix = np.sqrt(np.sum(d**2, axis=1))

    #finding the K nearest neighbors
    #an array of the indices of the K nearest distances/datapoints
    knearest = np.argsort(distance_matrix)[:K]
    
    #finding the K nearest labels
    klabels = np.zeros(K,) 
    for i in range(K):
        klabels[i] = y[int(knearest[i])]
                        
    #finding the most common label
    labels, counts = np.unique(klabels, return_counts=True)
    common_label = labels[np.argmax(counts)]
    return common_label
```
