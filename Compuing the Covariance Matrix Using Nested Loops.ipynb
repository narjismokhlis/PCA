This code I wrote uses a nested for loop to compute the covariance matrix of a given (N, D) array/dataset. Using vectorization gives faster results, and is more computationally economical. However, I thought of finding a way to compute the covariance matrix as fast as possible, while looping through entries. So this code exploits the symmetry of the covariance matrix, by only computing the variances of: 
- The diagonal entries where i = j
- The upper triangle entries [i, j] where i < j, and then storing the value in its "mirror entry" [j, i]. 
This way the code does not recompute the variance of the D(D - 1)/2 lower triangle entries

```python
import numpy as np

def cov(X):
    N, D = X.shape
    covariance = np.zeros(D, D)
    #creating a mean vector 
    mean = np.zeros(D)
    for n in range(N):
        mean += X[n]/N
    #Now we loop through the dataset to update the covariance matrix
    for n in range(N): #looping through the rows in the dataset
        for i in range(D): #looping vertically through the columns in the covariance matrix 
            for j in range(D): #looping horizontally through the column in the covariance matrix 
                #computing the variance in the diagonal entries where i==j
                if i==j:
                    #this is the formula for the variance of one variable 
                    covariance[i, j] += (1/N) * ((X[n, i] - mean[i])**2)
                #now we compute the covariance of two variables, but only the ones in the upper
                #triangle, which satisfy the condition i<j, and then we 
                # insert that value into the it's "mirror" entry [j, i]
                #remember: cov[x, y]=cov[y, x]
                elif i<j:
                    covariance[i, j] += (1/N) * ((X[n, i] - mean[i]) * (X[n, j] - mean[j]))
                    #inserting this value into it's mirror entry [j, i]
                    covariance[j, i] = covariance[i, j]

                else:
                    pass 
    return covariance     
```
